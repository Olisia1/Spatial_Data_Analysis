```python
x = sp.stats.uniform.rvs(0,10,50)
y = 0.5*x + sp.stats.uniform.rvs(-1,2,50)
data = pd.DataFrame(data = np.vstack([x,y]).T, columns=["x","y"])
data.head()
```

```python
fig, axes = plt.subplots(1,1,figsize=(5,5))
axes.scatter(x,y)
```

```python
fig, axes = plt.subplots(1,1,figsize=(5,5))
sns.scatterplot(data=data, x = "x", y = "y", ax = axes)
```

```python
fig, axes = plt.subplots(1,2,figsize=(10,5))
axes[0].scatter(x,y,color="red")
sns.scatterplot(data=data, x = "x", y = "y", color = "green", ax=axes[1])

```

```python
fig, axes = plt.subplots(2,2, figsize = (10,10))
axes[0,0].scatter(x,y,color="red")
axes[1,0].scatter(x,y,color="blue")
sns.scatterplot(data=data,x="x",y="y", ax = axes[0,1])
sns.scatterplot(data=data,x="x",y="y", ax= axes[1,1])
fig.suptitle("Tytul figury")

axes.set_title("Tytul wykresu")
axes.set_xlabel("opis x")
axes.set_ylabel("opis y")
axes.set_xlim([-2,10])
```
